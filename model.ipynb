```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, ExtraTreesClassifier ,AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from xgboost import XGBClassifier
```

```python
df.dropna(inplace=True)
df.drop_duplicates(inplace=True)
df.head()
```

```python
X = df.drop("Class",axis=1)
y = df["Class"]
y.unique()
```

```python
col = X.columns

scaler = StandardScaler()
X[col] = scaler.fit_transform(X[col])

X.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
results = []
models = {
    'Decision Tree_C': DecisionTreeClassifier(random_state=42),
    'Random Forest_C': RandomForestClassifier(random_state=42, n_jobs=-1),
    'Extra Trees_C': ExtraTreesClassifier(random_state=42, n_jobs=-1),
    'Gradient Boosting_C': GradientBoostingClassifier(random_state=42),
    'K-Neighbors_C': KNeighborsClassifier(),
    "XGBoost_C": XGBClassifier(random_state=42),
    "AdaBoost_C": AdaBoostClassifier(random_state=42),
}
```

```python
for name,model in models.items():
    model.fit(X_train,y_train)
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test,y_pred)
    results.append({
        "Name":name,
        "accuracy_score":acc,
    })

results_df = pd.DataFrame(results)
results_df
```

```python
final_model = RandomForestClassifier(random_state=42, n_jobs=-1)
final_model.fit(X_train,y_train)
y_pred = final_model.predict(X_test)
accuracy_score(y_test,y_pred)
```
